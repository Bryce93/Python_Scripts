```python
import random

def dnd_dice():
    """
    Prompts for dice details, lists each individual roll, 
    and provides the final sum.
    """
    # Converting input to integers
    try:
        faces = int(input("How many faces per die? (e.g., 20): "))
        number = int(input("How many dice are you rolling?: "))
    except ValueError:
        print("Please enter whole numbers only!")
        return

    # Generate a list of individual rolls
    # This uses a 'list comprehension' for efficiency
    rolls = [random.randint(1, faces) for _ in range(number)]
    
    total_roll = sum(rolls)
    
    print("-" * 20)
    print(f"Individual Rolls: {rolls}")
    print(f"Total Sum: {total_roll}")
    print("-" * 20)

dnd_dice()
```
