```python
import operator

class Calculator:
    def __init__(self):
        # Storing operations inside the class dictionary
        self.operations = {
            '+': (['addition', 'a', 'add'], operator.add),
            '-': (['subtraction', 's', 'sub'], operator.sub),
            '*': (['multiplication', 'm', 'multiply'], operator.mul),
            '/': (['division', 'divide', 'd'], operator.truediv),
            '**': (['exponent', 'power', 'p'], operator.pow)
        }

    def find_operation(self, user_input):
        user_input = user_input.strip().lower()
        for symbol, (aliases, func) in self.operations.items():
            if user_input == symbol.lower() or user_input in [a.lower() for a in aliases]:
                return symbol, aliases[0].capitalize(), func
        return None, None, None

    def run(self):
        print("=" * 40)
        print("        CLASS-BASED CALCULATOR PROJECT       ")
        print("=" * 40)
        
        while True:
            choice = input("Enter an operation symbol, name, or 'q' to quit: ")
            if choice.lower() == 'q':
                print("Goodbye!")
                break
                
            symbol, op_name, op_func = self.find_operation(choice)
            if not op_func:
                print("❌ Invalid operation. Try again.\n")
                continue
                
            try:
                num1 = float(input("Enter first number: "))
                num2 = float(input("Enter second number: "))
                
                if symbol == '/' and num2 == 0:
                    print("❌ Error: Division by zero.\n")
                    continue
                    
                result = op_func(num1, num2)
                print(f"Result ({op_name}): {num1} {symbol} {num2} = {result}\n")
            except ValueError:
                print("❌ Please enter numbers only.\n")

# 1. Create an instance of the class
if __name__ == "__main__":
    my_calc = Calculator()
    
    # 2. Call the method to start the app
    my_calc.run()
```
